```python
# CodeAlpha Task Automation
# JPG File Organizer

from pathlib import Path
import shutil


def organize_jpg_files():
    # Ask the user for the folder path
    folder_path = input(
        "Enter the path of the folder containing JPG files: "
    ).strip()

    source_folder = Path(folder_path)

    # Check if the folder exists
    if not source_folder.exists():
        print("Error: The folder does not exist.")
        return

    # Check if the path is a folder
    if not source_folder.is_dir():
        print("Error: The given path is not a folder.")
        return

    # Create destination folder
    destination_folder = source_folder / "JPG_Files"
    destination_folder.mkdir(exist_ok=True)

    moved_count = 0

    print("\nSearching for JPG files...\n")

    # Find JPG files
    for file in source_folder.iterdir():

        if file.is_file() and file.suffix.lower() == ".jpg":

            destination = destination_folder / file.name

            # Avoid overwriting existing files
            if destination.exists():
                print(f"Skipped: {file.name}")
                continue

            # Move the file
            shutil.move(str(file), str(destination))

            print(f"Moved: {file.name}")
            moved_count += 1

    # Display result
    print("\n================================")
    print("       AUTOMATION COMPLETE")
    print("================================")
    print(f"JPG files moved: {moved_count}")
    print(f"Destination folder: {destination_folder}")


# Start the program
if __name__ == "__main__":
    organize_jpg_files()
```
